```python
#Needed imports
import numpy as np
import tensorflow as tf
print(tf.__version__)
```

```python
#Creating the Input Layer
age = 34
experience = 7
input_layer = np.array([age, experience])
print(input_layer)
```

#Neural Network with 1 hidden layer

```python
#Creating the weights
weights_topnode = np.array([2,1])
weights_bottomnode = np.array([-2,-5])
output_weights = np.array([3,-1])
print(weights_topnode)
print(weights_bottomnode)
print(output_weights)
```

```python
#Calculating the Hidden Layer
top_node_value = (input_layer * weights_topnode).sum()
bottom_node_value = (input_layer * weights_bottomnode).sum()
print(top_node_value)
print(bottom_node_value)
```

```python
#Creating the Hidden Layer
hidden_layer_values = np.array([top_node_value, bottom_node_value])
print(hidden_layer_values)
```

```python
#Calculating the Prediction Layer
prediction = (hidden_layer_values * output_weights).sum()
print(prediction)
```

#Neural Network with 2 hidden layer with relu activation

```python
def relu(x):
  y=max(0,x)
  return y
```

```python
#Set first layer weights
weights_node_1 = np.array([2,1])
weights_node_2 = np.array([-2,-5])
```

```python
#Calculate first hidden layer with relu
n1 = (input_layer * weights_node_1).sum()
n2 = (input_layer * weights_node_2).sum()
n1_output = relu(n1)
n2_output = relu(n2)
layer_1_output = np.array([n1_output,n2_output])
print(layer_1_output)

```

```python
#Set second layer weights
weights_node_3 = np.array([3,1])
weights_node_4 = np.array([-1,-2])

```

```python
#Calculate second hidden layer with relu
n3 = (layer_1_output * weights_node_3).sum()
n4 = (layer_1_output * weights_node_4).sum()
n3_output = relu(n3)
n4_output = relu(n4)
layer_2_output = np.array([n3_output,n4_output])
print(layer_2_output)

```

```python
#Set output weights
output_weights = np.array([2,-6])

```

```python
#Calculate prediction
prediction = relu((layer_2_output * output_weights).sum())
print(prediction)

```

##Tensor Advance Math

```python
#Advance mathematical
x = tf.constant([-2.25, -48.23, 6.25])
print(tf.math.abs(x))
```

```python
#Exercise initialization
t1 = tf.constant([6, 4, 10])
t2 = tf.constant([3, 2, 4])

t3 = tf.Variable([[2.4, 2.5], [2.2, 1.6]])
t4 = tf.Variable([[9, 5], [16, 8]])

t_log = []
t_cumsum = []
t_tanh = []
t_sqrt = []
```

```python
def tensor_log(t,n):
  log = tf.math.log(tf.cast(t,tf.float32))
  t_log.append(f'{n} | {log}')

def tensor_cumsum(t, n):
  cumsum = tf.math.cumsum(t)
  t_cumsum.append(f'{n} | {cumsum}')

def tensor_tanh(t, n):
  tanh = tf.math.tanh(t)
  t_tanh.append(f'{n} | {tanh}')

def tensor_sqrt(t, n):
  sqrt = tf.math.sqrt(tf.cast(t,tf.float32))
  t_sqrt.append(f'{n} | {sqrt}')
```

```python
#Log a tensor
tensor_log(t1,"t1")
```

```python
#Print latest log
if len(t_log)!=0:
  print(t_log[-1])
else:
  print("No Log")
```

```python
#Print all logs
for index, item in enumerate(t_log):
  print(f'[{index}] {item}')
```

```python
#Cumsum a tensor
tensor_cumsum(t2,"t2")
```

```python
#Print latest cumsum
if len(t_cumsum)!=0:
  print(t_cumsum[-1])
else:
  print("No cumsum")
```

```python
#Print all cumsum
for index, item in enumerate(t_cumsum):
  print(f'[{index}] {item}')
```

```python
#Tahn a tensor
tensor_tanh(t3,"t3")
```

```python
#Print latest tanh
if len(t_tanh)!=0:
  print(t_tanh[-1])
else:
  print("No tahn")
```

```python
#Print all tanh
for index, item in enumerate(t_tanh):
  print(f'[{index}] {item}')
```

```python
#Sqrt a tensor
tensor_sqrt(t4,"t4")
```

```python
#Print latest sqrt
if len(t_sqrt)!=0:
  print(t_sqrt[-1])
else:
  print("No sqrt")
```

```python
#Print all sqrt
for index, item in enumerate(t_sqrt):
  print(f'[{index}] {item}')
```
